### Statistical Operations on Numerical Columns using NumPy

```python
# Identify numerical columns
numerical_cols = df.select_dtypes(include=['number']).columns

# Convert only numerical columns to a NumPy array for statistical operations
numerical_arr = df[numerical_cols].to_numpy()

print("Statistical operations for numerical columns:\n")

for i, col_name in enumerate(numerical_cols):
    print(f"Column: {col_name}")
    # Ensure the column is numeric before performing calculations
    # Although select_dtypes should handle this, an explicit check or conversion can be safer
    try:
        column_data = numerical_arr[:, i].astype(float)
        print(f"  Mean: {np.mean(column_data):.2f}")
        print(f"  Median: {np.median(column_data):.2f}")
        print(f"  Standard Deviation: {np.std(column_data):.2f}")
        print(f"  Minimum: {np.min(column_data):.2f}")
        print(f"  Maximum: {np.max(column_data):.2f}")
    except ValueError:
        print(f"  Could not perform statistical operations (non-numeric data found).")
    print("-" * 30)
```

### Bivariate Analysis

#### 1. Numerical-Numerical Relationships (Scatter Plots)

```python
# Example 1: Sales vs. Unit price
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Unit price', y='Sales', data=df, alpha=0.7)
plt.title('Sales vs. Unit Price')
plt.xlabel('Unit Price')
plt.ylabel('Sales')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
# Example 2: Sales vs. Quantity
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Quantity', y='Sales', data=df, alpha=0.7)
plt.title('Sales vs. Quantity')
plt.xlabel('Quantity')
plt.ylabel('Sales')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

#### 2. Numerical-Categorical Relationships (Box Plots)

```python
# Example 1: Sales by Branch
plt.figure(figsize=(10, 6))
sns.boxplot(x='Branch', y='Sales', data=df, palette='viridis')
plt.title('Sales Distribution by Branch')
plt.xlabel('Branch')
plt.ylabel('Sales')
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()
```

```python
# Example 2: Rating by Product line
plt.figure(figsize=(12, 7))
sns.boxplot(x='Product line', y='Rating', data=df, palette='plasma')
plt.title('Rating Distribution by Product Line')
plt.xlabel('Product Line')
plt.ylabel('Rating')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

#### 3. Categorical-Categorical Relationships (Count Plots)

```python
# Example 1: Gender distribution by Customer type
plt.figure(figsize=(10, 6))
sns.countplot(x='Customer type', hue='Gender', data=df, palette='coolwarm')
plt.title('Gender Distribution by Customer Type')
plt.xlabel('Customer Type')
plt.ylabel('Count')
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()
```

```python
# Example 2: City distribution by Branch
plt.figure(figsize=(10, 6))
sns.countplot(x='City', hue='Branch', data=df, palette='pastel')
plt.title('City Distribution by Branch')
plt.xlabel('City')
plt.ylabel('Count')
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()
```
